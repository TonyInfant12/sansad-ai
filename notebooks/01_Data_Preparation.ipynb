## Data Quality Finding

34 MPs have missing Attendance values.

The missing values are not random. Most correspond to MPs who were
ministers during the 17th Lok Sabha, while Om Birla is the Speaker.

Do not fill the missing Attendance values without a good reason.

For future analysis, Attendance should only be calculated among
MPs with reported attendance values.

```python
df["Start of Term"] = pd.to_datetime(
    df["Start of Term"],
    format="%d-%m-%Y"
)
```

## Date Conversion

The `Start of Term` column was converted from string format to
datetime format using `pd.to_datetime()`.

The original dates were in DD-MM-YYYY format. Explicit date formatting
was used to ensure the dates were parsed correctly.

This conversion allows date-based operations and calculations to be
performed later in the analysis.

```python
df["Start of Term"].dtype
```

```python
df["End of Term Date"] = pd.to_datetime(
    df["End of Term"].replace("In Office", pd.NaT),
    dayfirst=True
)
```

## End of Term Date Conversion

A new `End of Term Date` column was created from the original
`End of Term` column.

- Recorded dates were converted to datetime format.
- Representatives listed as `In Office` were represented as `NaT`
  because they do not have an end date in the dataset.
- The original `End of Term` column was retained to preserve the
  original information.
